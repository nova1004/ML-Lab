```python
import pandas as pd
import numpy as np

# Generate synthetic weather data
np.random.seed(42)
df = pd.DataFrame({
    'Humidity': np.random.uniform(30, 90, 100),
    'Pressure': np.random.uniform(1000, 1025, 100),
    'Temperature': np.random.uniform(15, 40, 100)
})
df.to_csv('weather_dataset.csv', index=False)
print("weather_dataset.csv created successfully.")
```

```python
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# Load data and split
df = pd.read_csv('weather_dataset.csv')
X_train, X_test, y_train, y_test = train_test_split(df[['Humidity', 'Pressure']], df['Temperature'], test_size=0.2, random_state=42)

# Train model
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

# Plot results
plt.figure(figsize=(8, 4))
plt.plot(y_test.values[:15], label='Actual Temp', marker='o')
plt.plot(predictions[:15], label='Predicted Temp', marker='x', linestyle='dashed')
plt.title('Weather Forecast: Actual vs Predicted')
plt.legend()
plt.show()
```
