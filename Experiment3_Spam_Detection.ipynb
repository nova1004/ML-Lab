```python
import pandas as pd
import random


spam_messages = [
    "WINNER! You have been selected for a $1000 cash prize. Click here.",
    "URGENT: Your bank account is locked. Verify your details immediately.",
    "Get 50% off your next prescription! No doctor required.",
    "Exclusive offer: Free vacation to the Bahamas. Call now to claim."
]

ham_messages = [
    "Hey, are we still meeting for lunch at the cafeteria?",
    "Can you send me the notes from the ML lecture?",
    "Happy birthday! Hope you have a great day.",
    "Don't forget to submit the lab report by tomorrow night."
]


data = []
for _ in range(50):
    data.append({'Message': random.choice(spam_messages), 'Category': 'spam'})
    data.append({'Message': random.choice(ham_messages), 'Category': 'ham'})

df = pd.DataFrame(data)


df = df.sample(frac=1, random_state=42).reset_index(drop=True)


df.to_csv('spam_dataset.csv', index=False)
print("spam_dataset.csv created successfully.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

df_spam = pd.read_csv('spam_dataset.csv')
df_spam['Category'] = df_spam['Category'].map({'ham': 0, 'spam': 1})

X_train, X_test, y_train, y_test = train_test_split(df_spam['Message'], df_spam['Category'], test_size=0.2, random_state=42)
vec = TfidfVectorizer(stop_words='english')
X_train_vec = vec.fit_transform(X_train)
X_test_vec = vec.transform(X_test)

model = MultinomialNB()
model.fit(X_train_vec, y_train)

predictions = model.predict(X_test_vec)
print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")
print("\nClassification Report:\n", classification_report(y_test, predictions))
```

```python
custom_messages = [
    "Hey Mitali, are we still meeting at the campus library tonight?",
    "CONGRATULATIONS! You have been selected to receive a $1000 gift card. Click here to CLAIM now!"
]
custom_vec = vec.transform(custom_messages)
results = model.predict(custom_vec)
for msg, pred in zip(custom_messages, results):
    label = "SPAM" if pred == 1 else "HAM (Normal)"
    print(f"Prediction: {label} | Message: '{msg}'\n")
```
